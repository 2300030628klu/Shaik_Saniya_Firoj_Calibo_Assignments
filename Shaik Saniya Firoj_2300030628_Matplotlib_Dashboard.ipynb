```python
import pandas as pd

data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],

    "name": ["Aarav", "Saniya", "Rahul", "Priya", "Arjun", "Sneha",
             "Vikram", "Ananya", "Kiran", "Meera", "Rohan", "Diya",
             "Aditya", "Fatima", "Varun", "Aisha", "Manoj", "Kavya",
             "Harsh", "Zoya"],

    "gender": ["Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male", "Female",
               "Male", "Female"],

    "department": ["CSE", "CSE", "ECE", "IT", "MECH", "CSE",
                   "ECE", "IT", "CSE", "ECE", "IT", "CSE",
                   "MECH", "IT", "ECE", "CSE", "MECH", "IT",
                   "CSE", "ECE"],

    "marks": [85, 92, 76, 88, 69, 95, 81, 74, 90, 87,
              72, 94, 78, 89, 83, 91, 67, 86, 88, 96],

    "study_hours": [4, 6, 3, 5, 2, 7, 4, 3, 6, 5,
                    3, 7, 4, 5, 4, 6, 2, 5, 5, 8]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 3, figsize=(15, 9))

# 1. Line Plot
axes[0, 0].plot(
    df["study_hours"],
    df["marks"],
    marker="o"
)
axes[0, 0].set_title("Study Hours vs Marks")
axes[0, 0].set_xlabel("Study Hours")
axes[0, 0].set_ylabel("Marks")


# 2. Bar Chart
department_avg = df.groupby("department")["marks"].mean()

axes[0, 1].bar(
    department_avg.index,
    department_avg.values
)
axes[0, 1].set_title("Average Marks by Department")
axes[0, 1].set_xlabel("Department")
axes[0, 1].set_ylabel("Average Marks")


# 3. Scatter Plot
axes[0, 2].scatter(
    df["study_hours"],
    df["marks"]
)
axes[0, 2].set_title("Study Hours vs Marks")
axes[0, 2].set_xlabel("Study Hours")
axes[0, 2].set_ylabel("Marks")


# 4. Histogram
axes[1, 0].hist(
    df["marks"],
    bins=8
)
axes[1, 0].set_title("Marks Distribution")
axes[1, 0].set_xlabel("Marks")
axes[1, 0].set_ylabel("Frequency")


# 5. Pie Chart
gender_count = df["gender"].value_counts()

axes[1, 1].pie(
    gender_count.values,
    labels=gender_count.index,
    autopct="%1.1f%%"
)
axes[1, 1].set_title("Gender Distribution")


# 6. Box Plot
axes[1, 2].boxplot(
    df["marks"]
)
axes[1, 2].set_title("Marks Box Plot")
axes[1, 2].set_ylabel("Marks")


plt.tight_layout()
plt.show()
```

```python


fig, ax = plt.subplots(figsize=(8, 5))

Department = ["CSE", "MECH", "IT", "ECE"]

colours = ["#065a82",
           "#1C7293",
           "#21295C",
           "#7FB3C7"]

for b, c in zip(Department, colours):

    sub = df[df.department == b]

    ax.scatter(
        sub.study_hours,
        sub.marks,
        label=b,
        color=c
    )

ax.legend(title="Branch")

ax.grid(alpha=0.3)

plt.savefig("c.png")

plt.show()
```

```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns





df = pd.read_csv("student_performance.csv")

print(df.head())
print(df.shape)
print(df.info())





plt.figure(figsize=(7, 4))

sns.histplot(
    data=df,
    x="Attendance",
    bins=8,
    kde=True
)

plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")
plt.grid(True)

plt.show()


```

```python
# 2. Study Hours vs Final Marks - Scatter Plot


plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Marks",
    hue="Gender",
    s=80
)

plt.title("Study Hours vs Final Marks")
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.legend(title="Gender")
plt.grid(True)

plt.show()
```

```python

# 3. Average Marks by Gender - Bar Chart

plt.figure(figsize=(7, 4))

sns.barplot(
    data=df,
    x="Gender",
    y="Final_Marks"
)

plt.title("Average Final Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Marks")
plt.grid(True, axis="y")

plt.show()
```

```python
# 4. Attendance vs Final Marks


plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Final_Marks",
    hue="Gender",
    s=80
)

plt.title("Attendance vs Final Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Marks")
plt.legend(title="Gender")
plt.grid(True)

plt.show()
```

```python
# 5. Subplot - All Four Visualizations


fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 10)
)


# ---- Plot 1: Histogram ----

sns.histplot(
    data=df,
    x="Attendance",
    bins=8,
    kde=True,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Attendance Distribution")
axes[0, 0].set_xlabel("Attendance (%)")
axes[0, 0].set_ylabel("Number of Students")
axes[0, 0].grid(True)


# ---- Plot 2: Study Hours vs Final Marks ----

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Marks",
    hue="Gender",
    s=70,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Study Hours vs Final Marks")
axes[0, 1].set_xlabel("Study Hours")
axes[0, 1].set_ylabel("Final Marks")
axes[0, 1].grid(True)
axes[0, 1].legend(title="Gender")


# ---- Plot 3: Average Marks by Gender ----

sns.barplot(
    data=df,
    x="Gender",
    y="Final_Marks",
    ax=axes[1, 0]
)

axes[1, 0].set_title("Average Final Marks by Gender")
axes[1, 0].set_xlabel("Gender")
axes[1, 0].set_ylabel("Average Final Marks")
axes[1, 0].grid(True, axis="y")


# ---- Plot 4: Attendance vs Final Marks ----

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Final_Marks",
    hue="Gender",
    s=70,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Attendance vs Final Marks")
axes[1, 1].set_xlabel("Attendance (%)")
axes[1, 1].set_ylabel("Final Marks")
axes[1, 1].grid(True)
axes[1, 1].legend(title="Gender")


# Adjust spacing
plt.tight_layout()

# Display all four graphs
plt.show()

```

```python
# 6. Correlation Analysis

print("\nCorrelation between Study Hours and Final Marks:")

study_correlation = df["Study_Hours"].corr(
    df["Final_Marks"]
)

print(study_correlation)


print("\nCorrelation between Attendance and Final Marks:")

attendance_correlation = df["Attendance"].corr(
    df["Final_Marks"]
)

print(attendance_correlation)
```

```python
# 7. Identify Insights


print("\n========== INSIGHTS ==========")

# Average marks by gender
gender_average = df.groupby("Gender")["Final_Marks"].mean()

print("\n1. Average Final Marks by Gender:")
print(gender_average)

# Average attendance
print("\n2. Average Attendance:")
print(df["Attendance"].mean())

# Average study hours
print("\n3. Average Study Hours:")
print(df["Study_Hours"].mean())

# Correlations
print("\n4. Study Hours vs Final Marks Correlation:")
print(study_correlation)

print("\n5. Attendance vs Final Marks Correlation:")
print(attendance_correlation)
```
